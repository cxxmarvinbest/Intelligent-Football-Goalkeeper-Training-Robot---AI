```python
video = cv2.VideoCapture("./test/14.mp4")
num = 0
save_step = 2
while True:
    ret, frame = video.read()
    if not ret:
        break
    num += 1
    if num % save_step == 0:
        cv2.imwrite("./demo_images14/" + str(num) + "(14).jpg" , frame)
```
